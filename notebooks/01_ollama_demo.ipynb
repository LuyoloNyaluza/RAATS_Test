```python
import subprocess
from langchain_community.llms import Ollama

def ollama_prompt(prompt):
             result = subprocess.run(
                 ["ollama", "run", "llama3", prompt],
                 capture_output=True, text=True, check=True, timeout=90 #seconds
             )
             return result.stdout.strip()

llm = Ollama(model="llama3")
print("LangChain Ollama response:")
print(llm.invoke("What is the sentiment of the word 'bullish'?"))

print("\nLangChain Ollama response 2:")
print(llm.invoke("Explain a moving average in one sentence."))
```
